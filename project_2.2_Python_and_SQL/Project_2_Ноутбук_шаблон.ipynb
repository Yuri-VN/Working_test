1. Напишите запрос, который позволит узнать, сколько (cnt) вакансий в каждом регионе (area).
Отсортируйте по количеству вакансий в порядке убывания.

```python
# текст запроса
# Здесь применил соединение 2-х таблиц по ключам используя оператор INNER JOIN. Слово INNER в операторе упустил, т.к. JOIN то тоже самое, что и INNER JOIN. 
query_4_1 = f'''select
    a.name as "region",
    count (v.id) as "cnt"
from
    public.vacancies v
    join public.areas a on v.area_id = a.id
group by region
order by cnt DESC
limit 10
    '''
```

# Юнит 6. Предметный анализ

1. Сколько вакансий имеет отношение к данным?

Считаем, что вакансия имеет отношение к данным, если в её названии содержатся слова 'data' или 'данн'.

*Подсказка: Обратите внимание, что названия вакансий могут быть написаны в любом регистре.*

6. Напишите запрос, позволяющий вычислить, какую зарплату для DS в **среднем** указывают для каждого типа требуемого опыта (уникальное значение из поля *experience*). 

При решении задачи примите во внимание следующее:
1. Рассматриваем только вакансии, у которых заполнено хотя бы одно из двух полей с зарплатой.
2. Если заполнены оба поля с зарплатой, то считаем зарплату по каждой вакансии как сумму двух полей, делённую на 2. Если заполнено только одно из полей, то его и считаем зарплатой по вакансии.
3. Если в расчётах участвует null, в результате он тоже даст null (посмотрите, что возвращает запрос select 1 + null). Чтобы избежать этой ситуацию, мы воспользуемся функцией [coalesce](https://postgrespro.ru/docs/postgresql/9.5/functions-conditional#functions-coalesce-nvl-ifnull), которая заменит null на значение, которое мы передадим. Например, посмотрите, что возвращает запрос `select 1 + coalesce(null, 0)`

Выясните, на какую зарплату в среднем может рассчитывать дата-сайентист с опытом работы от 3 до 6 лет. Результат округлите до целого числа. 

```python
# текст запроса
# Для выполнения условия по расчетусредней ЗП для DS, воспользовался конструкцией CASE... ELSE...END 
# Так же в случае NULL по признакам ЗП применил COALESCE (плюс при отсутствии обоих значений по ЗП подставляем 0). Далле уже знакомые фильтры.
query_6_6 = f'''select
    experience,
    ROUND(AVG
    (CASE 
    WHEN v.salary_from IS NOT NULL AND v.salary_to IS NOT NULL THEN (v.salary_from + v.salary_to) / 2.0
    ELSE COALESCE(v.salary_from, v.salary_to, 0)
    END), 0) as "AVG_Salary"
FROM 
    public.vacancies v
WHERE (v.name ILIKE '%data scientist%'
    or v.name ILIKE '%data science%' 
    or v.name ILIKE '%исследователь данных%'
    or v.name ILIKE '%machine learning%'
    or v.name ILIKE '%машинн%обучен%%' 
    or (v.name LIKE '%ML%' and v.name not ILIKE '%HTML%'))
    and (v.salary_from is not NULL or v.salary_to is not NULL)
group by v.experience   
    '''
```

```python
# результат запроса
df = pd.read_sql_query(query_6_6, connection)
df
```
